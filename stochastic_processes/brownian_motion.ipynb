```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
n_pasos = 50    # Número de iteraciones
dt = 0.1           # Diferencial de tiempo
sigma = 1.0        # Volatilidad / Coeficiente de difusión

# Generación de los incrementos aleatorios para X e Y
# Los pasos siguen una distribución normal con media 0 y varianza dt
dx = np.random.normal(0, sigma * np.sqrt(dt), n_pasos)
dy = np.random.normal(0, sigma * np.sqrt(dt), n_pasos)

# El movimiento es la suma acumulada de esos incrementos
# Iniciamos en el origen (0,0)
x = np.concatenate([[0], np.cumsum(dx)])
y = np.concatenate([[0], np.cumsum(dy)])

# Visualización
plt.figure(figsize=(8, 8))
plt.plot(x, y, lw=0.5, color='blue', alpha=0.8)

# Marcar el inicio y el fin
plt.scatter(x[0], y[0], color='green', label='Inicio (0,0)', zorder=5)
plt.scatter(x[-1], y[-1], color='red', label='Fin', zorder=5)

plt.title(f"Simulación de Movimiento Browniano en 2D ({n_pasos} pasos)")
plt.xlabel("Eje X")
plt.ylabel("Eje Y")
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.axis('equal') # Para que la escala sea la misma en ambos ejes
plt.savefig("movimiento_browniano_2D.png", dpi=300)
plt.show()
```
